# Bellabeat Smart Device Usage & Wellness Behaviour Analysis

## Stage 1: Ask

### Project Overview

Bellabeat is a wellness technology company that develops health-focused smart products.

The purpose of this case study is to analyse smart-device usage data to identify trends in consumer activity and wellness behaviour.

The findings will be used to develop high-level recommendations that could help inform Bellabeat's marketing strategy.

For this analysis, the Bellabeat app will be the primary product of focus.

## Business Task

The objective of this analysis is to examine smart-device usage data to identify meaningful trends in users' activity and wellness behaviours.

The analysis will investigate how consumers use fitness-tracking devices, identify behavioural patterns and relationships within the available data, and consider how these insights could be applied to Bellabeat app users.

The final goal is to translate the findings into high-level, data-informed recommendations that could support Bellabeat's marketing strategy.

## Business Questions

This analysis will address the following three primary questions:

1. What trends can be identified in smart-device usage?

2. How could these trends potentially apply to Bellabeat customers?

3. How could these trends help inform Bellabeat's marketing strategy?

## Product Focus: Bellabeat App

For this case study, the analysis will focus on the Bellabeat app.

The Bellabeat app provides users with health-related information including activity, sleep, stress, menstrual cycle and mindfulness data.

Because the available Fitbit dataset includes information related to physical activity and sleep, these behavioural dimensions can be used to explore how smart-device usage trends may be relevant to the Bellabeat app.

The analysis will not assume that Fitbit users are representative of Bellabeat customers. Instead, the dataset will be used as an exploratory source of consumer smart-device behaviour.

The completed case study will include:

2. A description of the data sources used.
3. Documentation of the data-cleaning and transformation process.
4. A summary of the analysis.

## Analytical Approach

The project will follow the six stages of the data analysis process:

### 1. Ask
Define the business problem, key stakeholders and analytical questions.

### 2. Prepare
Identify, collect, organise and assess the available data.

### 3. Process
Clean, validate and transform the data so it is ready for analysis.

### 4. Analyze
Explore the data, calculate summary statistics and identify meaningful trends and relationships.

### 5. Share
Communicate findings using clear visualisations and an interactive dashboard.

### 6. Act
Translate the findings into high-level recommendations that could support Bellabeat's marketing strategy.
