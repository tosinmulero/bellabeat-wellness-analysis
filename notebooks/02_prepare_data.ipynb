# Bellabeat Smart Device Usage & Wellness Behaviour Analysis

## Stage 2: Prepare

### Data Source

The primary dataset used for this analysis is the FitBit Fitness Tracker Data provided through Kaggle.

The dataset contains personal fitness tracker data from 30 eligible Fitbit users who consented to the submission of their data.

The available data includes information related to daily activity, steps, calories, intensity levels, heart rate and sleep.

For this project, the analysis will primarily use the data collected between 12 April 2016 and 12 May 2016.

### Data Organisation

The Fitbit dataset is organised into multiple CSV files containing different levels and types of fitness-tracker information.

The files include daily, hourly and minute-level records for activity, calories, intensity, steps, heart rate, sleep and weight.

For this project, the main files that will be used initially are:

- `dailyActivity_merged.csv`
- `sleepDay_merged.csv`

Additional files may be used later if they provide useful information for answering the business questions.

### Data Credibility and Limitations

The Fitbit dataset is useful for exploring patterns in smart-device usage, but it has several important limitations that should be considered when interpreting the results.

- The dataset contains data from only 30 Fitbit users, which is a relatively small sample.
- The users were Fitbit users rather than Bellabeat customers.
- The dataset covers a limited period of approximately one month.
- The available data may not represent the wider population of smart-device users.
- Some files may contain missing values or incomplete records.
- The dataset does not provide detailed demographic information such as age, gender, occupation or location.

Because of these limitations, the findings from this analysis will be treated as exploratory rather than fully representative of all Bellabeat customers or smart-device users.

### Data Integrity and Storage

The original Fitbit CSV files are stored in the `data/raw` folder and will not be modified directly.

A separate `data/processed` folder will be used later to store cleaned and transformed datasets created during the analysis.

Before analysis, the data will be checked for:

- Missing values
- Duplicate records
- Incorrect data types
- Invalid or unusual values
- Inconsistent date and time formats
- Differences in the number of users across files

These checks will help confirm that the data is suitable for analysis and will provide a clear record of the data-cleaning process.
