### Daily Activity Data Cleaning Summary

The `dailyActivity_merged.csv` dataset initially contained 940 rows and 15 columns.

The following data-quality checks were completed:

- Confirmed 33 unique Fitbit user IDs.
- Converted `ActivityDate` from text to datetime format.
- Confirmed that there were no missing values.
- Confirmed that there were no exact duplicate rows.
- Verified that the dataset covers 12 April 2016 to 12 May 2016.
- Identified that users contributed between 4 and 31 daily records, indicating unequal data coverage across participants.
- Checked the main numeric activity variables for negative values and found none.
- Identified 4 records where `Calories` was equal to 0.
- These records also showed no meaningful recorded activity and were treated as incomplete tracker records.
- Removed the 4 zero-calorie records from the cleaned analytical dataset.

The cleaned daily activity dataset contains 936 rows and 15 columns.

The original raw CSV file remains unchanged.

### Sleep Data Cleaning Summary

The `sleepDay_merged.csv` dataset initially contained 413 rows and 5 columns.

The following data-quality checks were completed:

- Converted `SleepDay` from text to datetime format.
- Confirmed there were no missing values.
- Identified 3 exact duplicate records.
- Removed the 3 duplicate records.
- Confirmed 24 unique Fitbit users have sleep data.
- Identified that users contributed between 1 and 31 sleep records.
- Verified that the sleep data covers 12 April 2016 to 12 May 2016.
- Inspected unusually short sleep records and retained them because they remained logically consistent.
- Confirmed that `TotalMinutesAsleep` never exceeded `TotalTimeInBed`.
- Confirmed that each remaining user-date combination is unique.

The cleaned sleep dataset contains 410 rows and 5 columns.

The original raw CSV file remains unchanged.
